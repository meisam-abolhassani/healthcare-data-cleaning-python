## Task 2 — Data Cleaning

### 2.1 Standardizing the Admission Date

```python
naive = pd.to_datetime(df['Admission_Date'], dayfirst=True, errors='coerce')

pd.DataFrame({
    'Patient_ID': df['Patient_ID'],
    'Original': df['Admission_Date'].astype(str),
    'Naive_Result': naive
})
```

```python
from datetime import datetime

def fix_admission_date(value):
    """
    Two cases exist in this column.

    Case 1 - the value is already a datetime.
    Excel read the original text DD-MM-YY as YY-MM-DD,
    so the day and the year must be swapped back.

    Case 2 - the value is still text.
    Parsed with dayfirst=True to respect the DD-MM-YYYY order.
    """
    if isinstance(value, datetime):
        day   = value.year - 2000
        month = value.month
        year  = 2000 + value.day
        return pd.Timestamp(year=year, month=month, day=day)

    return pd.to_datetime(str(value).strip(), dayfirst=True, format='mixed')
```

```python
df['Admission_Date'] = df['Admission_Date'].apply(fix_admission_date)

pd.DataFrame({
    'Patient_ID': df['Patient_ID'],
    'Original': df_raw['Admission_Date'].astype(str),
    'Cleaned': df['Admission_Date'].dt.strftime('%Y-%m-%d')
})
```

### 2.4 Correcting Invalid Email Addresses

```python
def fix_email(value):
    """
    Normalizes an address and repairs a domain with no suffix.

    The domain is only modified when it contains no dot at all,
    so a valid address is never touched.
    """
    email = str(value).strip().lower()

    if '@' not in email:
        return email

    local, _, domain = email.partition('@')

    if '.' not in domain:
        domain = domain + '.com'

    return local + '@' + domain
```

```python
email_pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'

email_before = df['Email'].copy()
df['Email'] = df['Email'].apply(fix_email)

pd.DataFrame({
    'Patient_ID': df['Patient_ID'],
    'Before': email_before,
    'Valid_Before': email_before.astype(str).str.match(email_pattern),
    'After': df['Email'],
    'Valid_After': df['Email'].str.match(email_pattern)
})
```

### 2.5 Handling the Missing Contact Number

```python
def fix_contact(value):
    """
    A phone number is an identifier, not a measurement,
    so the missing entry is labeled rather than estimated.
    """
    if pd.isna(value):
        return 'Not Provided'

    return str(int(value))
```

```python
df['Contact_Number'] = df['Contact_Number'].apply(fix_contact)

pd.DataFrame({
    'Patient_ID': df['Patient_ID'],
    'Before': df_raw['Contact_Number'],
    'After': df['Contact_Number']
})
```

### 2.6 Imputing the Missing Blood Pressure Reading

```python
import numpy as np

group_systolic  = df.groupby('Diagnosis')['Systolic_BP'].transform('mean').astype('float')
group_diastolic = df.groupby('Diagnosis')['Diastolic_BP'].transform('mean').astype('float')

systolic_filled  = df['Systolic_BP'].astype('float').fillna(group_systolic)
diastolic_filled = df['Diastolic_BP'].astype('float').fillna(group_diastolic)

# Adding 0.5 before flooring forces a trailing .5 upward.
# Python rounds it to the nearest even number instead, so round(142.5) returns 142.
df['Systolic_BP']  = np.floor(systolic_filled + 0.5).astype('Int64')
df['Diastolic_BP'] = np.floor(diastolic_filled + 0.5).astype('Int64')

print("Blood pressure imputed using the diagnosis group mean")
```

```python
print("Missing contact numbers:", (df['Contact_Number'] == 'Not Provided').sum())
print("Missing systolic  :", df['Systolic_BP'].isnull().sum())
print("Missing diastolic :", df['Diastolic_BP'].isnull().sum())
print()

df[['Patient_ID', 'Name', 'Diagnosis', 'Systolic_BP', 'Diastolic_BP', 'Contact_Number']]
```

### 3.2 Standardizing the Gender Column

```python
print("Unique values:", df['Gender'].unique())
print("Trailing or leading spaces:", (df['Gender'] != df['Gender'].str.strip()).sum())
print()

df['Gender'].value_counts()
```

```python
gender_map = {
    'm': 'Male',   'male': 'Male',     'man': 'Male',   'boy': 'Male',
    'f': 'Female', 'female': 'Female', 'woman': 'Female', 'girl': 'Female'
}

def standardize_gender(value):
    """
    Maps known spellings to a single canonical form.

    A value that is not in the map is returned cleaned rather
    than discarded, so an unexpected category stays visible
    instead of being silently absorbed into Unknown.
    """
    if pd.isna(value):
        return 'Unknown'

    key = str(value).strip().lower()

    if key in gender_map:
        return gender_map[key]

    return str(value).strip().title()
```

```python
test_values = pd.Series(['male', ' FEMALE ', 'M', 'f', 'Male', 'woman', None, 'Non-binary'])

pd.DataFrame({
    'Input': test_values.astype(str),
    'Output': test_values.apply(standardize_gender)
})
```

```python
gender_before = df['Gender'].copy()
df['Gender'] = df['Gender'].apply(standardize_gender)

print("Records modified:", (gender_before != df['Gender']).sum())
print("Unique values after:", df['Gender'].unique())
print()

for column in ['Gender', 'Diagnosis', 'Medication', 'Doctor_Assigned']:
    print(f"{column:18} {sorted(df[column].unique())}")
```

### 3.4 Renaming Columns to a Consistent snake_case Convention

```python
import re
def to_snake_case(name):
    """
    Converts a column label to snake_case without a hardcoded map.

    Three passes:
      1. spaces and dashes become underscores
      2. an acronym followed by a word is split, so BPValue becomes BP_Value
      3. a lowercase or digit followed by a capital is split, so heartRate becomes heart_Rate
    """
    name = name.strip()
    name = re.sub(r'[\s\-]+', '_', name)
    name = re.sub(r'(?<=[A-Z])(?=[A-Z][a-z])', '_', name)
    name = re.sub(r'(?<=[a-z0-9])(?=[A-Z])', '_', name)
    name = re.sub(r'_+', '_', name)
    return name.lower()

test_labels = ['Patient_ID', 'Doctor Assigned', 'admissionDate', 'BPValue', '  Heart-Rate  ']

pd.DataFrame({
    'Input': test_labels,
    'Output': [to_snake_case(label) for label in test_labels]
})
```

```python
columns_before = list(df.columns)
df.columns = [to_snake_case(column) for column in df.columns]

pd.DataFrame({
    'Before': columns_before,
    'After': list(df.columns)
})
```
